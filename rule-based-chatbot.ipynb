```python
from datetime import datetime

print("Chatbot started! Type 'bye' to exit. \n")

while True: 
    user = input("you:").lower().strip()

    if any(w in user.split() for w in ["hello", "hi", "hey"]):
       print("Bot: Hello! Welcome.")
    elif "how are you" in user:
        print("Bot: I'm fine, thank you")
    elif "your name" in user:
        print("Bot: Myself Bot!, nice to meet you")
    elif "time" in user:
        print("Bot:", datetime.now().strftime("It's %I:%M %p."))
    elif "date" in user:
        print("Bot:", datetime.now().strftime("Today is %d %B %Y."))
    elif "bye" in user or "exit" in user:
        print("Bot: Goodbye! See you next time.")
        break
    else:
        print("Bot: Sorry, I don't understand that. Please try something else.")
        
```
